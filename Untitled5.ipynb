```python
import numpy as np, torch, torchvision, random, json, os
from sklearn.model_selection import train_test_split

ROLL = "M26ECM505"                       # <-- put YOUR roll number here
SEED = int("".join(c for c in ROLL if c.isdigit()))
print("Seed =", SEED)                    # M26ECC001 -> 26001

random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)                 # should say cuda
```

```python
train_full = torchvision.datasets.CIFAR10(root="./data", train=True, download=True)
test_set   = torchvision.datasets.CIFAR10(root="./data", train=False, download=True)

X_full = train_full.data                 # (50000, 32, 32, 3) uint8
y_full = np.array(train_full.targets)    # (50000,)
X_test = test_set.data                   # (10000, 32, 32, 3)
y_test = np.array(test_set.targets)
print(X_full.shape, X_test.shape)
```

```python
idx = np.arange(len(y_full))
train_idx, val_idx = train_test_split(
    idx, test_size=5000, stratify=y_full, random_state=SEED)

os.makedirs("data", exist_ok=True)
np.savez("data/saved_split.npz", train_idx=train_idx, val_idx=val_idx, seed=SEED)

print(len(train_idx), len(val_idx))                  # 45000 5000
print(np.bincount(y_full[val_idx]))                  # 500 per class
```

```python
mean = X_full[train_idx].reshape(-1,3).mean(0)/255
std  = X_full[train_idx].reshape(-1,3).std(0)/255   # statistics from TRAIN only

def to_tensor(X):
    X = torch.tensor(X, dtype=torch.float32).permute(0,3,1,2)/255
    return (X - torch.tensor(mean).view(1,3,1,1).float()) / torch.tensor(std).view(1,3,1,1).float()

Xtr, ytr = to_tensor(X_full[train_idx]), torch.tensor(y_full[train_idx])
Xva, yva = to_tensor(X_full[val_idx]),   torch.tensor(y_full[val_idx])
Xte, yte = to_tensor(X_test),            torch.tensor(y_test)
print(Xtr.shape, Xva.shape, Xte.shape)
```

```python
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

def evaluate(model, X, y, bs=1000):
    model.eval(); correct=0; loss=0
    with torch.no_grad():
        for i in range(0, len(X), bs):
            xb, yb = X[i:i+bs].to(device), y[i:i+bs].to(device)
            out = model(xb)
            loss += nn.functional.cross_entropy(out, yb, reduction="sum").item()
            correct += (out.argmax(1)==yb).sum().item()
    return loss/len(X), correct/len(X)

def train(model, Xtr, ytr, Xva, yva, epochs=30, lr=1e-3, bs=128, wd=0.0):
    model.to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=wd)
    dl = DataLoader(TensorDataset(Xtr, ytr), batch_size=bs, shuffle=True)
    hist = {"tr_loss":[], "tr_acc":[], "va_loss":[], "va_acc":[]}
    for ep in range(epochs):
        model.train()
        for xb, yb in dl:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad()
            nn.functional.cross_entropy(model(xb), yb).backward()
            opt.step()
        tl, ta = evaluate(model, Xtr, ytr)
        vl, va = evaluate(model, Xva, yva)
        for k, v in zip(hist, [tl, ta, vl, va]): hist[k].append(v)
        print(f"ep {ep+1:02d} train {ta:.3f} val {va:.3f}")
    return hist

def count_params(m): return sum(p.numel() for p in m.parameters())
```

```python
os.makedirs("logs", exist_ok=True)
Xtr, ytr, Xva, yva, Xte, yte = [t.to(device) for t in (Xtr, ytr, Xva, yva, Xte, yte)]

def make_models():
    torch.manual_seed(SEED)
    mlp_A = nn.Sequential(nn.Flatten(), nn.Linear(3072,512), nn.ReLU(),
                          nn.Linear(512,256), nn.ReLU(), nn.Linear(256,10))
    mlp_B = nn.Sequential(nn.Flatten(), nn.Linear(3072,1024), nn.ReLU(),
                          nn.Linear(1024,512), nn.ReLU(), nn.Linear(512,10))
    cnn = nn.Sequential(
        nn.Conv2d(3,64,3,padding=1), nn.ReLU(), nn.MaxPool2d(2),
        nn.Conv2d(64,128,3,padding=1), nn.ReLU(), nn.MaxPool2d(2),
        nn.Conv2d(128,256,3,padding=1), nn.ReLU(), nn.MaxPool2d(2),
        nn.Flatten(), nn.Linear(256*4*4,320), nn.ReLU(), nn.Linear(320,10))
    return {"MLP_A": mlp_A, "MLP_B": mlp_B, "CNN": cnn}

models = make_models()
for k, m in models.items():
    print(k, count_params(m))
```

```python
import copy, json

def train_best(model, name, epochs=20, lr=1e-3, bs=128):
    model.to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    hist = {"tr_loss":[], "tr_acc":[], "va_loss":[], "va_acc":[]}
    best_va, best_state, best_ep = -1, None, 0
    for ep in range(epochs):
        model.train()
        perm = torch.randperm(len(Xtr), device=device)
        for i in range(0, len(Xtr), bs):
            b = perm[i:i+bs]
            opt.zero_grad()
            nn.functional.cross_entropy(model(Xtr[b]), ytr[b]).backward()
            opt.step()
        tl, ta = evaluate(model, Xtr, ytr)
        vl, va = evaluate(model, Xva, yva)
        for k, v in zip(hist, [tl, ta, vl, va]): hist[k].append(v)
        if va > best_va:
            best_va, best_ep = va, ep+1
            best_state = copy.deepcopy(model.state_dict())
        print(f"{name} ep {ep+1:02d} train {ta:.3f} val {va:.3f}")
    model.load_state_dict(best_state)
    hist["best_epoch"] = best_ep
    json.dump(hist, open(f"logs/{name}.json", "w"))
    return hist
```

```python
def predict(model, X, bs=2000):
    model.eval(); out = []
    with torch.no_grad():
        for i in range(0, len(X), bs):
            out.append(model(X[i:i+bs]).argmax(1))
    return torch.cat(out)

def make_noisy_labels(y_true, frac, seed):
    g = torch.Generator().manual_seed(seed)
    y = y_true.cpu().clone()
    n = len(y)
    idx = torch.randperm(n, generator=g)[:int(frac*n)]
    y[idx] = torch.randint(0, 10, (len(idx),), generator=g)
    return y.to(device)
```

```python
def run_noise(frac, name, epochs=100, lr=1e-3, bs=128):
    model = make_models()["MLP_B"].to(device)      # same init as before
    y_noisy = make_noisy_labels(ytr, frac, SEED)
    corrupt = (y_noisy != ytr)                     # labels that are actually wrong
    clean = ~corrupt
    print(name, "corrupted fraction:", corrupt.float().mean().item())

    opt = torch.optim.Adam(model.parameters(), lr=lr)
    H = {k: [] for k in ["clean_acc", "memorised", "true_on_corrupt", "val_acc"]}
    best_va, best_state, best_ep = -1, None, 0

    for ep in range(epochs):
        model.train()
        perm = torch.randperm(len(Xtr), device=device)
        for i in range(0, len(Xtr), bs):
            b = perm[i:i+bs]
            opt.zero_grad()
            nn.functional.cross_entropy(model(Xtr[b]), y_noisy[b]).backward()
            opt.step()

        p = predict(model, Xtr)
        H["clean_acc"].append((p[clean] == ytr[clean]).float().mean().item())
        H["memorised"].append((p[corrupt] == y_noisy[corrupt]).float().mean().item())
        H["true_on_corrupt"].append((p[corrupt] == ytr[corrupt]).float().mean().item())
        va = (predict(model, Xva) == yva).float().mean().item()
        H["val_acc"].append(va)
        if va > best_va:
            best_va, best_ep = va, ep+1
            best_state = copy.deepcopy(model.state_dict())
        if (ep+1) % 10 == 0:
            print(f"ep {ep+1:3d} clean {H['clean_acc'][-1]:.3f} memorised {H['memorised'][-1]:.3f} "
                  f"true@corrupt {H['true_on_corrupt'][-1]:.3f} val {va:.3f}")

    final_test = (predict(model, Xte) == yte).float().mean().item()
    model.load_state_dict(best_state)
    best_test = (predict(model, Xte) == yte).float().mean().item()
    H["best_epoch"] = best_ep; H["test_at_best_val"] = best_test; H["test_at_end"] = final_test
    json.dump(H, open(f"logs/{name}.json", "w"))
    return H

H40  = run_noise(0.4, "q1b_noise40")
H100 = run_noise(1.0, "q1b_random100")
```

```python
def run_random_subset(n=10000, epochs=200, lr=1e-3, bs=128):
    model = make_models()["MLP_B"].to(device)
    g = torch.Generator().manual_seed(SEED)
    sub = torch.randperm(len(Xtr), generator=g)[:n].to(device)
    Xs = Xtr[sub]
    ys = torch.randint(0, 10, (n,), generator=g).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    H = {"fit_acc": [], "val_acc": []}
    for ep in range(epochs):
        model.train()
        perm = torch.randperm(n, device=device)
        for i in range(0, n, bs):
            b = perm[i:i+bs]
            opt.zero_grad()
            nn.functional.cross_entropy(model(Xs[b]), ys[b]).backward()
            opt.step()
        H["fit_acc"].append((predict(model, Xs) == ys).float().mean().item())
        H["val_acc"].append((predict(model, Xva) == yva).float().mean().item())
        if (ep+1) % 20 == 0:
            print(f"ep {ep+1:3d} fit-random-labels {H['fit_acc'][-1]:.3f} val {H['val_acc'][-1]:.3f}")
    json.dump(H, open("logs/q1b_random_subset.json", "w"))
    return H

Hsub = run_random_subset()
plt.plot(Hsub["fit_acc"], label="train (random labels)"); plt.plot(Hsub["val_acc"], label="val (true labels)")
plt.xlabel("epoch"); plt.ylabel("accuracy"); plt.legend(); plt.savefig("logs/q1b_random_subset.png", dpi=200); plt.show()
```

```python
import torch, torch.nn as nn, torch.nn.functional as F, numpy as np
dev = 'cuda' if torch.cuda.is_available() else 'cpu'

# ---------- transforms (fixed, applied identically to train/val/test) ----------
def to_gray(X):
    g = (0.299*X[:,0] + 0.587*X[:,1] + 0.114*X[:,2]).unsqueeze(1)
    return g.repeat(1,3,1,1)

def make_patch_perm(seed, grid=4):                 # 4x4 grid of 8x8 patches
    g = torch.Generator().manual_seed(seed)
    return torch.randperm(grid*grid, generator=g)

def patch_shuffle(X, perm, grid=4):
    N,C,H,W = X.shape; p = H//grid
    patches = X.unfold(2,p,p).unfold(3,p,p)         # N,C,g,g,p,p
    patches = patches.reshape(N,C,grid*grid,p,p)[:,:,perm]
    patches = patches.reshape(N,C,grid,grid,p,p).permute(0,1,2,4,3,5)
    return patches.reshape(N,C,H,W)

# ---------- generic train / eval ----------
def train_model(model, Xtr,ytr,Xva,yva, epochs=20, lr=1e-3, bs=128, wd=5e-4):
    model.to(dev); opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=wd)
    best, best_state = 0, None
    for ep in range(epochs):
        model.train(); idx = torch.randperm(len(Xtr))
        for i in range(0,len(Xtr),bs):
            b = idx[i:i+bs]; x,y = Xtr[b].to(dev), ytr[b].to(dev)
            opt.zero_grad(); F.cross_entropy(model(x),y).backward(); opt.step()
        va = accuracy(model,Xva,yva)
        if va>best: best, best_state = va, {k:v.clone() for k,v in model.state_dict().items()}
        print(f'ep {ep+1:02d} val {va:.3f}')
    model.load_state_dict(best_state); return model

@torch.no_grad()
def predict(model, X, bs=1000):
    model.eval(); out=[]
    for i in range(0,len(X),bs): out.append(F.softmax(model(X[i:i+bs].to(dev)),1).cpu())
    return torch.cat(out)

def accuracy(model,X,y): return (predict(model,X).argmax(1)==y).float().mean().item()

def conf_analysis(probs, y, n_bins=10):
    conf, pred = probs.max(1); correct = (pred==y)
    ece = 0
    bins = torch.linspace(0,1,n_bins+1)
    for lo,hi in zip(bins[:-1],bins[1:]):
        m = (conf>lo)&(conf<=hi)
        if m.any(): ece += m.float().mean()*abs(correct[m].float().mean()-conf[m].mean())
    return dict(acc=correct.float().mean().item(), mean_conf=conf.mean().item(),
                conf_correct=conf[correct].mean().item(),
                conf_wrong=conf[~correct].mean().item() if (~correct).any() else float('nan'),
                ece=float(ece))

# ---------- run both conditions for both models ----------
perm = make_patch_perm(seed=SEED)                   # SEED = your roll-number seed
conds = {'gray': to_gray, 'patch': lambda X: patch_shuffle(X, perm)}
results = {}
for cname, T in conds.items():
    for mname, ctor in [('CNN',CNN), ('MLP_A',MLP_A)]:
        torch.manual_seed(SEED)
        m = train_model(ctor(), T(Xtr),ytr, T(Xva),yva)
        probs = predict(m, T(Xte))
        results[(mname,cname)] = conf_analysis(probs, yte)
        # per-class accuracy
        pc = [(probs.argmax(1)[yte==c]==c).float().mean().item() for c in range(10)]
        results[(mname,cname)]['per_class'] = pc
        print(mname, cname, results[(mname,cname)])
```

```python
import torch, torch.nn.functional as F

@torch.no_grad()
def predict(model, X, bs=1000):
    model.eval(); out = []
    for i in range(0, len(X), bs):
        out.append(F.softmax(model(X[i:i+bs].to(dev)), 1).cpu())
    return torch.cat(out)

def accuracy(model, X, y):
    return (predict(model, X).argmax(1) == y.cpu()).float().mean().item()

def conf_analysis(probs, y, n_bins=10):
    y = y.cpu()
    conf, pred = probs.max(1); correct = (pred == y)
    ece = 0.0
    bins = torch.linspace(0, 1, n_bins+1)
    for lo, hi in zip(bins[:-1], bins[1:]):
        m = (conf > lo) & (conf <= hi)
        if m.any():
            ece += m.float().mean().item() * abs(correct[m].float().mean().item() - conf[m].mean().item())
    return dict(acc=correct.float().mean().item(),
                mean_conf=conf.mean().item(),
                conf_correct=conf[correct].mean().item(),
                conf_wrong=conf[~correct].mean().item() if (~correct).any() else float('nan'),
                ece=ece)

perm = make_patch_perm(seed=SEED)
conds = {'gray': to_gray, 'patch': lambda X: patch_shuffle(X, perm)}
results = {}
yte_cpu = yte.cpu()

for cname, tf in conds.items():                 # tf instead of T, so your global T is not overwritten
    for mname, ctor in [('CNN', make_cnn), ('MLP_A', make_mlp_a)]:
        torch.manual_seed(SEED)
        m = train_model(ctor(), tf(Xtr), ytr, tf(Xva), yva)
        probs = predict(m, tf(Xte))
        res = conf_analysis(probs, yte)
        pred = probs.argmax(1)
        res['per_class'] = [(pred[yte_cpu == c] == c).float().mean().item() for c in range(10)]
        results[(mname, cname)] = res
        print(mname, cname, res)
```

```python
import torch, torch.nn as nn, torch.nn.functional as F
dev = 'cuda' if torch.cuda.is_available() else 'cpu'

class ConvAE(nn.Module):
    def __init__(self, d):                          # d = bottleneck size
        super().__init__()
        self.enc = nn.Sequential(
            nn.Conv2d(3,32,3,2,1), nn.ReLU(),       # 16x16
            nn.Conv2d(32,64,3,2,1), nn.ReLU(),      # 8x8
            nn.Conv2d(64,128,3,2,1), nn.ReLU(),     # 4x4
            nn.Flatten(), nn.Linear(128*4*4, d))
        self.dec_fc = nn.Linear(d, 128*4*4)
        self.dec = nn.Sequential(
            nn.ReLU(),
            nn.ConvTranspose2d(128,64,4,2,1), nn.ReLU(),
            nn.ConvTranspose2d(64,32,4,2,1), nn.ReLU(),
            nn.ConvTranspose2d(32,3,4,2,1), nn.Sigmoid())
    def encode(self, x): return self.enc(x)
    def forward(self, x):
        z = self.enc(x)
        return self.dec(self.dec_fc(z).view(-1,128,4,4))

@torch.no_grad()
def ae_mse(ae, Xin, Xtgt, bs=1000):
    ae.eval(); s = 0.0
    for i in range(0, len(Xin), bs):
        s += F.mse_loss(ae(Xin[i:i+bs].to(dev)), Xtgt[i:i+bs].to(dev), reduction='sum').item()
    return s / Xtgt.numel()

def train_ae(d, Xin_tr, Xtgt_tr, Xin_va, Xtgt_va, epochs=20, lr=1e-3, bs=128, noise_fn=None):
    ae = ConvAE(d).to(dev)
    opt = torch.optim.Adam(ae.parameters(), lr=lr)
    best, state = 1e9, None
    for ep in range(epochs):
        ae.train(); idx = torch.randperm(len(Xtgt_tr))
        for i in range(0, len(idx), bs):
            b = idx[i:i+bs]; tgt = Xtgt_tr[b].to(dev)
            inp = noise_fn(tgt) if noise_fn else Xin_tr[b].to(dev)   # fresh noise every step for the DAE
            opt.zero_grad(); F.mse_loss(ae(inp), tgt).backward(); opt.step()
        v = ae_mse(ae, Xin_va, Xtgt_va)
        if v < best: best, state = v, {k: x.clone() for k, x in ae.state_dict().items()}
        print(f'  d={d} ep {ep+1:02d} val MSE {v:.5f}')
    ae.load_state_dict(state)
    return ae, best
```

```python
import copy, torch, torch.nn as nn
dev = "cuda" if torch.cuda.is_available() else "cpu"

def get_model(x):
    if isinstance(x, (tuple, list)): x = x[0]
    if isinstance(x, dict): x = x.get("model", list(x.values())[0])
    return x

def get_encoder(ae):
    if hasattr(ae, "encoder"): return ae.encoder
    if hasattr(ae, "encode"):  return ae.encode
    raise AttributeError("no encoder/encode found; run print(model)")

@torch.no_grad()
def feats_lin(enc, X, bs=1000):
    if isinstance(enc, nn.Module): enc.eval().to(dev)
    return torch.cat([enc(X[i:i+bs].to(dev)).flatten(1).cpu()
                      for i in range(0, len(X), bs)])

def probe_lin(Ftr, ytr, Fva, yva, Fte, yte, wds=(0, 1e-4, 1e-3, 1e-2), epochs=100, seed=0):
    mu, sd = Ftr.mean(0), Ftr.std(0) + 1e-6          # train statistics only
    Ftr, Fva, Fte = [(f - mu) / sd for f in (Ftr, Fva, Fte)]
    Ftr_d, ytr_d = Ftr.to(dev), ytr.to(dev)
    best = None
    for wd in wds:                                    # choose wd on validation
        torch.manual_seed(seed)
        clf = nn.Linear(Ftr.shape[1], 10).to(dev)
        opt = torch.optim.Adam(clf.parameters(), lr=1e-3, weight_decay=wd)
        for _ in range(epochs):
            perm = torch.randperm(len(Ftr_d), device=dev)
            for i in range(0, len(Ftr_d), 512):
                idx = perm[i:i+512]
                loss = nn.functional.cross_entropy(clf(Ftr_d[idx]), ytr_d[idx])
                opt.zero_grad(); loss.backward(); opt.step()
        with torch.no_grad():
            va = (clf(Fva.to(dev)).argmax(1).cpu() == yva.cpu()).float().mean().item()
        if best is None or va > best[0]:
            best = (va, wd, clf)
    va, wd, clf = best
    with torch.no_grad():                             # test evaluated once
        te = (clf(Fte.to(dev)).argmax(1).cpu() == yte.cpu()).float().mean().item()
    return dict(val=round(va, 4), test=round(te, 4), wd=wd)

models_q3 = {f"AE{k}": get_model(aes[k]) for k in aes}
models_q3["DAE1024"] = get_model(dae)

probe_results = {}
for name, m in models_q3.items():
    enc = get_encoder(m)
    probe_results[name] = probe_lin(feats_lin(enc, Xtr), ytr, feats_lin(enc, Xva), yva,
                                    feats_lin(enc, Xte), yte)
    print(name, probe_results[name])

# baseline 1: raw pixels
flat = lambda X: X.flatten(1).float()
probe_results["RawPixels"] = probe_lin(flat(Xtr), ytr, flat(Xva), yva, flat(Xte), yte)
print("RawPixels", probe_results["RawPixels"])

# baseline 2: random untrained d=1024 encoder
torch.manual_seed(0)
rand_ae = copy.deepcopy(models_q3["AE1024"])
for mod in rand_ae.modules():
    if hasattr(mod, "reset_parameters"): mod.reset_parameters()
renc = get_encoder(rand_ae)
probe_results["RandomEnc1024"] = probe_lin(feats_lin(renc, Xtr), ytr, feats_lin(renc, Xva), yva,
                                           feats_lin(renc, Xte), yte)
print("RandomEnc1024", probe_results["RandomEnc1024"])

import json; json.dump(probe_results, open("q3_probe_results.json", "w"), indent=2)
print({n: sum(p.numel() for p in m.parameters()) for n, m in models_q3.items()})
```
